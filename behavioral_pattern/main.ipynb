### **Behavioral Design Pattern**
Behavioral patterns are where engineers start thinking like system designers, not just coders.

Behavioral pattern answer: <br>
"How do objects communicate and share responsibility?"

They focus on: <br>
- Loose coupling 
- Clean flow of control
- Extensible behavior without rewriting code

### **Observer Pattern**<br>

Problem it solves <br>
You have:<br>
- One object (Subject) whose state changes.
- Many dependent objects (Observers) that must react.

But: <br>
- You don't want tight coupling
- You don't want to hardcode calls everywhere

**Real-life analogy** <br>
👉 YouTube channel<br>
- Channel uploads video
- Subscribers get notified automatically.

**Backend Example** <br>
When: <br>
- Order is placed
- Payment is successful

You want: <br>
- Email notification
- Analytics update
- Audit log

**Other Services**

```python
class EmailService:
    def send(self):
        print("Email notification sent successfully!")

class AnalyticsService:
    def track(self):
        print("Analytics service updated!")
```

**Bad approach (tight coupling)**

```python
class OrderService:
    def __init__(self):
        self.email_service = EmailService()
        self.analytics_service = AnalyticsService()
    def place_order(self):
        print("Order placed")
        self.email_service.send()
        self.analytics_service.track()
```

Every new listener → modify code

### **Observer Pattern**

step-1: Observer Interface

```python
class Observer:
    def update(self):
        raise NotImplementedError
```

step-2: Subject

```python
class OrderService:
    def __init__(self):
        self.observers = []
    
    def subscribe(self, observer):
        self.observers.append(observer)
    
    def notify(self, data):
        for observer in self.observers:
            observer.update(data)
        
    def place_order(self):
        print("Order placed")
        self.notify({"status": "ORDERED_PLACED"})
```

step-3: Concrete Observers

```python
class EmailObservice(Observer):
    def update(self, data):
        print("Email sent:", data)

class AnalyticsService(Observer):
    def update(self, data):
        print("Analytics updated", data)
```

**Usage**

```python
order = OrderService()

order.subscribe(EmailObservice())
order.subscribe(AnalyticsService())

order.place_order()
```

**Staff Engineer mindset**<br>
Observer is used in: <br>
- Django signals <br>
- Event-driven systems <br>
- Message queues (Kafka, RabbitMQ)

👉 Core pattern for scalable systems

### **Strategy Pattern** <br>

Problem it solves <br>

You have: <br>
- Multiple ways to do the same task <br>
- Want to switch behavior at runtime <br>
- Want to avoid if-else hell

**Real-life analogy** <br>
👉 Google Maps <br>
- Driving 
- Walking
- Cycling

Same destination, different strategies.

**Backend example**

Payment methods: <br>
* Credit Card
* PayPal
* Stripe

**Bad approach**

```python
def pay(amount, method):
    if method == "card": 
        pass 
    elif method == "paypal":
        pass
```

Hard to extend! If you want to add new payment type "Bkash" then you need to modify pay function what will violate Open/Close principal.

**Strategy Pattern**

step-1: Strategy interface

```python
class PaymentStrategy:
    def pay(self, amount):
        raise NotImplementedError
```

step-2: Concrete Strategies

```python
class CardPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid ${amount} by card")

class PaypalPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid ${amount} via Paypal")
```

step-3: Context

```python
class PaymentService:
    def __init__(self, strategy: PaymentStrategy):
        self.strategy = strategy
    
    def checkout(self, amount):
        self.strategy.pay(amount)
```

**Usage**

```python
service = PaymentService(CardPayment())
service.checkout(100)

service = PaymentService(PaypalPayment())
service.checkout(200)
```

**Staff Engineer mindset**

Strategy is: <br>

- One of the MOST asked interview patterns <br>
- Used in validation logic, pricing, routing

👉 Essential for clean architecture

### **Command Pattern**

Problem it solves

You want to:
- Encapsulate a request as an object
- Queue it
- Log it
- Undo/redo it

**Real-life analogy** <br>
👉 Remote Control <br>
- Button = command
- TV doesn't know who passed it

**Backend example** <br>
* Job queue
* Task execution
* API action handling

| Pattern  | Solves               | Common Use        |
| -------- | -------------------- | ----------------- |
| Observer | Event notification   | Signals, events   |
| Strategy | Algorithm selection  | Payments, pricing |
| Command  | Action encapsulation | Jobs, tasks       |
