**L — Liskov Substitution Principle (LSP)**

Definition: "If class B (Child) is a subclass of class A (Parent), then you should be able to replace A (Parent) with B (Child) without breaking the program."

In simple words: <br>
- Child class must honor the contract of the parent class
- No surprises
- No broken behavior

Why **LSP** matters in backend systems? <br>

Backend systems rely heavily on: <br>
- Interfaces
- Abstractions
- Dependency injection

Breaking **LSP** leads to: <br>
- Runtime bugs
- Weird edge cases
- Broken APIs

**Classic LSP Violation (Easy Example)**

Wrong Design

```python
class Bird:
    def fly(self):
        print("Flying")

class Penguin(Bird):
    def fly(self):
        raise Exception("Penguins can't fly")
```

Why this breaks LSP?

```python
def make_bird_fly(bird: Bird):
    bird.fly()

make_bird_fly(Penguin()) # 💥 crash
```

You can not substitute (replace) Bird with Penguin.

👉 Inheritance lie detected

**Correct Design**

```python
class Bird:
    pass 

class FlyingBird(Bird):
    def fly(self):
        print("Flying")

class Penguin(Bird):
    def swim(self):
        print("Swimming")
```

✔ Now substitution is safe

**Backend-Oriented Example 1: File Storage Service**

Very common in backend systems.

**Bad Design (LSP Violation)**

```python
# Interface
class FileStorage:
    def save(self, file):
        raise NotImplementedError

    def get_public_url(self, file):
        raise NotImplementedError
```

**S3 Implementation**

```python
class S3Storage(FileStorage):
    def save(self, file):
        print("File saved to s3")
    
    def get_public_url(self, file):
        return f"https://s3.amazonaws.com/file/{file}"
```

**Local Storage**

```python
class LocalStorage(FileStorage):
    def save(self, file):
        print("File saved locally")
    
    def get_public_url(self, file):
        # since local storage has no public url so raise Exception
        raise Exception("Local storage has no public URL")
```

**Usage**

```python
s3 = S3Storage()
s3.save("shamim.txt")
s3.get_public_url("shamim.txt")
```

```python
local = LocalStorage()
local.save("shamim.txt")
local.get_public_url("shamim.txt")
```

**Correct Design (Split responsibilities - SRP + LSP)**

```python
class Storage:
    def save(self, file):
        raise NotImplementedError

class PublicStorage(Storage):
    def get_public_url(self, file):
        raise NotImplementedError
```

Implementations

```python
class S3Storage(PublicStorage):
    def save(self, file):
        print(f"{file} saved to S3")
    
    def get_public_url(self, file):
        return f"https://s3.amazon.com/file/{file}"

class LocalStorage(Storage):
    def save(self, file):
        print(f"{file} saved to local storage successfully!") 
```

**Usage**

```python
s3 = S3Storage()
s3.save("shamim.txt")
s3.get_public_url("shamim.txt")
```

```python
local = LocalStorage()
local.save("shamim.txt")
```

✔ LSP preserved <br>
✔ No fake behavior

**Backend-Oriented Example 2: Authentication Strategy**

Using Strategy Pattern.

**LSP Violation**

```python
class AuthStrategy:
    def authenticate(self, request):
        raise NotImplementedError 
```

```python
class TokenAuth(AuthStrategy):
    def authenticate(self, request):
        return True
```

```python
class SessionAuth(AuthStrategy):
    def authenticate(self, request):
        raise Exception("Session Expired!")
```

**Whey broken**

Beause caller expect `True` or `False` but it also gets `Exception` from SessionAuth what breaks the program.

Contract Broken!

**Correct Design (Follow LSP)**

```python
class AuthStrategy:
    def authenticate(self, request) -> bool:
        raise NotImplementedError

class TokenAuth(AuthStrategy):
    def authenticate(self, request):
        return True 

class SessionAuth(AuthStrategy):
    def authenticate(self, request):
        return False 
```

✔ Same return type <br>
✔ Same expectations <br>
✔ Safe substitution

**Backend-Oriented Example 3: Notification Observer**

Using Observer Pattern.

**Violate LSP**

```python
class Observer:
    def update(self, event):
        raise NotImplementedError

class EmailObserver(Observer):
    def update(self, event):
        print(f"{event}")

# Violate LSP
class SMSObserver(Observer):
    def update(self, event):  
        raise Exception("SMS service down")
```

```python
observers = [EmailObserver(), SMSObserver()]

def notify_all(observers, event):
    for observer in observers:
        observer.update(event)

notify_all(observers, "Happy New Year!")
```

Email Service was fine but SMS service broken the program which is the violation of LSP. Because here SMS Service (Observer) just break the publisher (notify_all function)

**Correct Design**

```python
class SMSObserver(Observer):
    def update(self, event):
        try:
            raise Exception("Service Down")
        except Exception:
            print("SMS failed but handled.")
```

```python
observers = [EmailObserver(), SMSObserver()]

def notify_all(observers, event):
    for observer in observers:
        observer.update(event)

notify_all(observers, "Happy New Year!")
```

✔ Observers never break the publisher

**Key LSP Rules (Very Important)**

A subclass must not:
- Strengthen preconditions
- Weaken postconditions
- Change return types
- Throw unexpected exceptions

**LSP Smells (Memorize these)**

If a subclass:
- Raises NotImplementedError
- Throws unexpected exceptions
- Says “this doesn’t apply here”
- Has empty or dummy methods

👉 You probably violated LSP
