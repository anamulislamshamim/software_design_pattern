**Categories of Objects in Design** <br>

There are generally three categories of Objects in Software Design <br>
- Entity Objects
- Boundary Objects
- Control Objects

**Entity Objects** <br>

It is the most familiar, because they correspond to some real-world entity in the problem space. If you have an Object representing a chair in your software, then this is an entity object. If you have an object representing a building or a customer, these are all entity objects. Generally, these objects will know attributes about themselves. They will also be able to modify themselves, and have some rules for how to do so.

When you are identifying objects to include in  your software and breaking down into smaller objects, then you will initially get entity objects. The other categories of objects will come later, as you start to think about the technical design of the software.

**Entity vs Value Object**

**Value Object (Postal Address)**

```python
# Value Object
from dataclasses import dataclass

@dataclass(frozen=True)
class PostalAddress:
    street: str
    city: str 
    zip_code: str 

    def __eq__(self, other):
        if not isinstance(other, PostalAddress):
            return False
        return (
            self.street == other.street and 
            self.city == other.city and 
            self.zip_code == other.zip_code
        )
```

**The Entity Object (User)**

This Object has a unique user_id. Even if two users have the same name, they are not the same person if their IDs differ.

```python
class User:
    def __init__(self, user_id: str, name: str, address: PostalAddress):
        self.user_id = user_id
        self.name = name 
        self.address = address 
    
    def __eq__(self, other):
        if not isinstance(other, User):
            return False 
        # # Equality is based ONLY on the identity (ID)
        return self.user_id == other.user_id
    
    def update_address(self, new_address: PostalAddress):
        """Entities are mutable; we can change their state."""
        self.address = new_address
```

**Comparing the Behavior** <br>

Entity object properties is mutable but the unique identity is always same through the life cycle. That means if two object contain same properties and value and unique id is different. They are completely two different Object. Entity Object's unique identiy will always preserved.

On the other side, Value Object is immutable. If properties change that means it is new Object. If contains same properties with same value that means they are same Object

```python
addr1 = PostalAddress("123 Maple St", "Springfield", "62704")
addr2 = PostalAddress("123 Maple St", "Springfield", "62704")

# Value Objects are equal if their data matches
print(f"Addresses are equal: {addr1 == addr2}")  # Output: True

# Create two User Entities with the same name but different IDs
user1 = User(user_id="USR-001", name="Alice", address=addr1)
user2 = User(user_id="USR-002", name="Alice", address=addr1)

# Entities are NOT equal if their IDs are different, even if data matches
print(f"Users are equal: {user1 == user2}")      # Output: False
```

See though user1 and user1 has same name still they are different.

**Boundary Objects** <br>

In software design—particularly within the Entity-Control-Boundary (ECB) pattern—a Boundary Object acts as the interface between the system and the outside world.

If the Entity is the "data" and the Control is the "logic," the Boundary is the "translator." It handles how the system communicates with actors, whether those actors are human users, external APIs, or hardware devices.

**Core Responsibilities** <br>
- Translation: It converts external input (like a mouse click or a JSON payload) into a format the internal system understands.

- Validation: It performs basic checks on incoming data before passing it to the logic layers.

- Isolation: It protects the internal "business logic" from changes in the external environment. If you switch your UI from a Web page to a Mobile app, only the Boundary objects should change.

**Real-World Example: A Banking ATM** <br>
Imagine you are withdrawing cash from an ATM.

1. The Actor: You (the customer).

2. The Boundary Object: The ATM Screen/Keypad. This object captures your PIN and the amount you want. It doesn't know how to verify your identity; it just collects the data and shows you the result.

3. The Control Object: The Withdrawal Coordinator. This is the "brain" that checks if you have enough money and tells the machine to dispense cash.

3. The Entity Object: Your Bank Account. This is the record that holds your balance and transaction history.

Python Example: A Web API Boundary
In modern web development, a Controller or a Request Schema often acts as the Boundary.

```python
from abc import abstractmethod
import uuid

class RegistrationService:
    @abstractmethod
    def register(username, password):
        user = User(username, password)
        return user


# The Boundary Object 
# It handles the outside world (HTTP request)

class UserRegistrationBoundary:
    def handle_request(self, json_data):
        #  1. Captulare input
        username = json_data.get("username")
        password = json_data.get("password")

        # 2. Simple validation (Boundary logic)
        if not username or len(password) < 8:
            return {"status": 400, "message": "Invalid Input"} 
        
        # 3. Pass to the "Control" object (the logic)
        result = RegistrationService.register(username, password)
        return result.__dict__


# The Entity Object
# Pure data and identity

class User:
    def __init__(self, username, password):
        self.user_id = str(uuid.uuid4())
        self.username = username
        self.password = password
```

```python
registration = UserRegistrationBoundary()
print(registration.handle_request({"username": "Shamim", "password": "dksk499dks"}))
```

**Types of Boundary Objects** <br>

You will encounter Boundary objects in three main forms:

- User Interfaces (UI): Screens, buttons, and command-line prompts.

- System Interfaces: APIs, Gateways, and Wrappers that talk to other software (like a Stripe Payment Gateway).

- Device Interfaces: Drivers that talk to hardware (like a printer or a sensor).

**The Control Object (The "Verb")** <br>
