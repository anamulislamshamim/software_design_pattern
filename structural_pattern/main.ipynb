### **Structural Design Pattern**
This pattern answer this following question:<br>
"How do we connect and organize classes/objects without making the system fragile?"<br>
They focus on:<br>
* Composition over inheritance
* Decoupling 
* Clean integration with external systems.

#### **Adapter Pattern**
Problem it solves: <br>
You want to use an existing class, but:<br>
* Its interface does **NOT** match what your system expects.
* You can not change that class (third-party, legacy code)

**Real-life Analogy**<br>
* your laptop expects 3 point socket but wall socket is 2 point
* So, use Adapter to convert 2 point to 3 points.

**Backend Example**<br><br>
You have:
* Your app expects `send(message)`
* A third-party service provides `push(text)`

**Without Adapter**

```python
class ThirdPartySMS:
    def push(self, text):
        print(f"SMS sent: {text}")
    
def notify(service):
    service.push("Hello") # tightly coupled

# Usage 
service = ThirdPartySMS()
notify(service)
```

**With Adapter Pattern**

```python
# Step-1: Expected Interface
class NotificationService:
    def send(self, message):
        raise NotImplementedError
    
# Step-2: Third-party class 
class ThirdPartySMS:
    def push(self, text):
        print(f"SMS sent: {text}")

# Adapter
class SMSAdapter(NotificationService):
    def __init__(self, service_provider):
        self.service_provider = service_provider
    
    def send(self, message):
        self.service_provider.push(message)
```

**Usage**

```python
sms_provider = ThirdPartySMS()

adapter = SMSAdapter(sms_provider)

adapter.send("Hi! Shamim...")
```

**Staff Engineer mindset**<br>

Use Adapter when:<br>
* Integrating third-party APIs<br>
* Migrating legacy systems<br>
* Writing clean abstractions

#### **Decorator Pattern**
Problem it solves <br>
You want to:<br>
* Add new behavior 
* Without modifying existing code
* Without creating too many subclasses

**Real-life analogy**<br>
Coffee:<br>
* Plain coffee <br>
* Coffee + milk <br>
* Coffee + sugar <br>
* Coffee + milk + sugar

You don't create new class for each combination.

**Backend Example**<br>
You have: <br>
* A base API handler 
* Want to add logging, auth, caching dynamically

##### Bad approach (inheritance explosion)

```python
class LoggedService:
    pass 

class AuthLoggedService:
    pass 

class CachedAuthLoggedService:
    pass 
```

**inheritance nightmare**

#### **Decorator Pattern**

```python
# Step-1: Base interface
class Service:
    def execute(self):
        raise NotImplementedError


# Step-2: Core implementation
class CoreService(Service):
    def execute(self):
        print("Executing core service")


# Step-3: Decorator Base
class ServiceDecorator(Service):
    def __init__(self, service: Service):
        self.service = service

    def execute(self):
        return self.service.execute()


# Step-4: Concrete Decorators
class LoggingDecorator(ServiceDecorator):
    def execute(self):
        print("Logging before execution...")
        return super().execute()


class AuthDecorator(ServiceDecorator):
    def execute(self):
        print("Checking authentication...")
        return super().execute()
```

**Usage**

```python
# Client code
service = CoreService()
service = LoggingDecorator(service)
service = AuthDecorator(service)

service.execute()
```

**Staff Engineer mindset**<br>

Decorator is used everywhere:<br>
* Django middleware
* Python decorators (@login_required)
* API request pipelines

👉 One of the MOST important patterns

### **Facade Pattern**
Problem it solves <br>
A system is too complex:<br>
* Many classes  <br>
* Many steps <br>
* Hard to use correctly.

You want:
- **One Simple Interface**

**Real-life Analogy** <br>
👉 Restaurant waiter <br>

You don't talk to: <br>
- Chef
- Cleaner
- Cashier

You talk to **one waiter**.

**Backend Example** <br>

Placing an order involves: <br>
- Payment service
- Inventory service
- Notification service

Without Facade (complex)

```python
class PaymentService:
    pass 

class InventoryService:
    pass 

class NotificationService:
    pass 

# Usage
payment = PaymentService()
inventory = InventoryService()
notification = NotificationService()

payment.pay()
inventory.reserve()
notification.send()
```

Every caller must know everything!


**Facade Pattern**

```python
# Subsystems 
class PaymentService:
    def pay(self):
        print("Payment processed")

class InventoryService:
    def reserve(self):
        print("Inventory reserved")
    
class NotificationService:
    def send(self):
        print("Notification sent")
```

**Facade**

```python
class OrderFacade:
    def __init__(self):
        self.payment = PaymentService()
        self.inventory = InventoryService()
        self.notification = NotificationService()
    
    def place_order(self):
        self.payment.pay()
        self.inventory.reserve()
        self.notification.send()
```

**Usage**

```python
order = OrderFacade()
order.place_order()
```

**Staff Engineer mindset** <br>

Facade is excellent for:
- Service orchestration
- Use-case based APIs
- Clean controllers/views

👉 Very common in microservices orchestration layers

**Quick Comparison** <br>


| Pattern   | Purpose                           | Used When              |
| --------- | --------------------------------- | ---------------------- |
| Adapter   | Make incompatible interfaces work | Third-party APIs       |
| Decorator | Add behavior dynamically          | Logging, auth, caching |
| Facade    | Simplify complex systems          | Use-case orchestration |
